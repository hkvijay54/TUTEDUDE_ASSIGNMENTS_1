```python
'''2. Apply the following discount rules and print the final amount: 
order amount 2000 15% discount 
             1500 <= order amount < 2000 10% discount 
             1000 <= order amount < 1500 7% discount 
otherwise 0% discount '''

order_amount = int(input("Enter you order amount: "))
if order_amount >= 2000 :
    print ("you got 15% discount ")
    print ("total amount to pay : ", order_amount-(order_amount*(15/100)))
elif order_amount >= 1500 and order_amount < 2000:
    print ("you got 10% discount ")
    print ("total amount to pay : ", order_amount-(order_amount*(10/100)))
elif order_amount >= 1000 and order_amount < 1500:
    print ("you got 7% discount ")
    print ("total amount to pay : ", order_amount-(order_amount*(7/100)))
else:
    print("Sorry no discount \n your total amount is: ",order_amount)
```

```python
'''3. Ensure you convert Input to a number and handle the case when the input is non 
numeric by displaying an error message and exiting the program.'''

order_amount =input("Enter you order amount: ")
try:
    order_amount=float(order_amount)
    if order_amount >= 2000 :
        print ("you got 15% discount ")
        print ("total amount to pay : ", order_amount-(order_amount*(15/100)))
    elif order_amount >= 1500 and order_amount < 2000:
        print ("you got 10% discount ")
        print ("total amount to pay : ", order_amount-(order_amount*(10/100)))
    elif order_amount >= 1000 and order_amount < 1500:
        print ("you got 7% discount ")
        print ("total amount to pay : ", order_amount-(order_amount*(7/100)))
    else:
        print("Sorry no discount \n your total amount is: ",order_amount)
except ValueError:
    print("Please enter the valid number")
```

####  Task 2: Process Multiple Orders (for loop) 

```python
'''1. Given a list of order amounts: orders = [1200, 2500, 800, 1750, 3000], use a for loop 
to apply the discount rules from Task 1 to each order and print a summary table 
showing: order_amount -> discount% -> final_amount_ 
'''
orders=[1200, 2500, 800, 1750, 3000]
print(f"{'order_amount':20}{'discount':15}{'total_amount':15}")
print("-"*50)
for i in orders:
    if i >= 2000:
        discount=15
    elif 1500 <= i <2000:
        discount = 10
    elif 1000 <= i <1500:
        discount=7
    else:
        discount=0
    total= i - (i*discount/100)
    print(f'{i:10}{discount:15}{total:15}')
```

```python
# 2.Also compute and print the total revenue after discounts.
orders=[1200, 2500, 800, 1750, 3000]
print(f"{'order_amount':20}{'discount':15}{'total_amount':15}")
print("-"*50)
total_rev=0
for i in orders:
    if i >= 2000:
        discount=15
    elif 1500 <= i <2000:
        discount = 10
    elif 1000 <= i <1500:
        discount=7
    else:
        discount=0
    total= i - (i*discount/100)
    print(f'{i:10}{discount:15}{total:15}')
    total_rev+=total
print("-"*50)
print(f'{'total_revenue':20}{total_rev:20}')
```

##### Task 3: User Menu (while loop + break/continue) 

```python
'''1. Implement a simple while-loop driven menu that keeps asking the user to choose an 
action until they press q to quit. 
Menu options: 
1 — Add order amount to a running list (use input()) 
2 — Show all orders and totals after applying discounts 
q — Quit'''
orders=[]
while True:
    print("Menu")
    print("-"*20)
    print("1.Add \n 2.show \n q.quit(press q)")
    choice=input('choose you action : ')
    if choice == '1':
        i=input('enter to add')
        orders.append(i)
    elif choice == '2':
        print('orders :',orders)
    elif choice == 'q':
        print("exited, thank you!")
        break
    else:
        print('invalid input')
        break
```

### Task 5: Loop Control with Conditions (break & continue) 

```python
'''1. Given a list of daily sales: daily - [200, 150, 0, 400, 50, -1, 300], iterate through it with 
a for loop and: 
If a day's value is -1 , treat it as corrupted data and break the loop (stop 
processing). 
If a day's value is 0, treat it as a day with no sales and continue (skip adding to 
revenue). 
For valid positive sales, add to total_sales and print the running total. 
'''

daily_sales=[200, 150, 0, 400, 50, -1, 300]
total_sales = 0
for i in daily_sales:
    if i == -1:
        print("correupted data.. Exiting")
        break
    elif i == 0:
        print('No sales today')
        continue
    else:
        total_sales =total_sales + i
    print(f'added {i} Running total of the day: ',total_sales)
```

```python
'''2 Print the final total after the loop completes (or stops due to corrupted data). '''
daily_sales=[200, 150, 0, 400, 50, -1, 300]
total_sales = 0
for i in daily_sales:
    if i == -1:
        print("correupted data.. Exiting")
        break
    elif i == 0:
        print('No sales today')
        continue
    else:
        total_sales =total_sales + i
print('Final total: ',total_sales)
```
