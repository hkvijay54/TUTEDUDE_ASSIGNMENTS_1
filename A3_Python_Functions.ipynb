```python
# Task 1 - Basic Function: Price After Discount 
def apply_discount(price, discounted_percent=5):
    if discounted_percent > 60:
        print ("=====Discount upto 60% only======")
        discounted_percent = 60
    discounted_price= price - (price * discounted_percent /100)
    return discounted_percent, discounted_price

disc_price, result_1 = apply_discount(1000, 10)
print(f"Price after {disc_price}% discount: {result_1}")

disc_price, result_2 = apply_discount(500)
print(f"Price after {disc_price}% discount: {result_2}")

disc_price, result_3 = apply_discount(2000, 65)
print(f"Price after {disc_price}% discount: {result_3}")

```

```python
# Task 2 - Recursive Function: Factorial utility
def factorial(n):
    if n<0:
        print('factorial is not defined.exit...')
        return None
    if n==0 or n==1:
        return 1
    return n* factorial(n-1)
res1=factorial(5)
print(f'factorial (5)= {res1}')

res2=factorial(0)
print(f'factorial (0)= {res2}')

res3=factorial(-3)
print(f'factorial (-3)= {res3}')
```

```python
#Task 3 - Lambda Function: GST Calculator 
gst=lambda price:price +(0.18 * price) #18% gst
price=int(input('Enter price : '))
print(f'total price is: {price} and price after gst is {gst(price)}')

#compute gst price on after discounted price

gst_discount_price=lambda price,discounted_percent: (price - (price * discounted_percent/100))*1.18
print (f'Price is with discount and gst : {gst_discount_price(price,10)}')

print (f'Price is with discount and gst : {gst_discount_price(price,50)}')
```

```python
#Task 4 - using map(): Apply GST to List of Prices
prices=[100,250,400,1200,50]
gst=lambda price:price +(0.18 * price) #18% gst

prices_gst=list(map(gst,prices))

print(f"Original prices:  {prices}")
print(f"Prices after GST: {prices_gst}")
```

```python
#Task 5 - using filter(): Filter Expensive Products
prices = [100, 250, 400, 1200, 50, 2000, 850]

#prices greater than 500
expensive=list(sorted(filter(lambda price:price > 500, prices)))

#prices less than equal than 500
budget=list(sorted(filter(lambda price:price <= 500, prices)))

print('original product prices:', prices)

print('Expensive product prices: ',expensive)

print('budget product prices',budget)
```

```python
#Task 6 - Combined utility Function 
def process_prices(prices):
    #10% discount to all prices
    discounted_prices=list(map(lambda price: price - (price * .10),prices))
    #keep only discounted prices that are above 300
    filter_price=list(filter(lambda price:price > 300,discounted_prices))
    return discounted_prices, filter_price

dis_prices=[500, 900, 50, 750]
all_discounted, filter_price =process_prices(dis_prices)
print ('original prices: ',dis_prices)
print("discounted prices are: ",all_discounted)
print('prices more than 300 are: ',filter_price)
```

```python
#Task 7 - Mini Problem: Menu using Functions 
def add_price(prices_list,price):
    prices_list.append(price)
    print (f'{price} sucessfuly added to price_list')

def get_average_price(prices_list):
    if not prices_list:
        return 0
    return sum(prices_list)/len(prices_list)

def get_max_price(prices_list):
    if not prices_list:
        return 0
    return max(prices_list)

lis_prices=[]
while True:
    print('========Billing==========\n')
    print('1. Add price')
    print('2. Get Average price')
    print('3. Get max price')
    print('q. Exit')

    choice = input('Enter your choice: ').strip().lower()

    if choice == '1':
        user_price=int(input('Enter the product price : '))
        add_price(lis_prices,user_price)

    elif choice == '2':
        average=get_average_price(lis_prices)
        print("The average prices of product purchase is : ",average)
    
    elif choice == '3':
        max_price=get_max_price(lis_prices)
        print('The max product price is: ',max_price)
    
    elif choice == 'q':
        print('Exiting..... ok Byee !!')
        break

    else:
        print('Enter invalid input, Enter 1,2,3,q')

```
